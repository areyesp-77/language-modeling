# Importar librerías 🇱 y módulos  🧰

Las librerías son conjuntos de código ya escrito que pueden reutilizarse para realizar diferentes tareas. La librería puede contener varios módulos, que son pequeños programas que realizan tareas más pequeñas o especializadas dentro de la librería, por ejemplo, la librería es Office y uno de sus módulos es Word.

Antes de usar la librería o algún módulo, hay que instalar. Si se usa el IDLE de Python o la terminal, lo común es usar el comando PIP:


```
# pip install LIBRARY
```

Cuando se usa algún editor, es muy probable que éste haga el proceso de instalación de forma automática. Para ello, primero debe existir una llamada en el código que se está escribiendo:

```
# import LIBRARY
```

Cuando esto sucede, aparecerá un mensaje para indicar que hay un error porque Python no encuentra la librería o el módulo. Por lo general, se resuelve dando click en el nombre de la librería o módulo para que el editor proceda a la instalación.

Captura de pantalla 2026-03-03 a la(s) 12.59.54 p.m..png

Una vez que la librería se ha instalado, se puede llamar a la librería completa o sólo a uno de sus módulos:

```
# import LIBRARY
```

```
# from LIBRARY import MODULE
```

Adicionalmente, es muy común que estos códigos se importen mediane abreviaturas que son comunes entre la gente que conoce el lenguaje:

```
# import LIBRARY as ABBR
```

```
# import pandas as pd
```

Cuando se usa la abreviatura, ya no es necesario usar el nombre de la librería o el módulo, sino sólo la abreviatura:

```
# x = pd.read ('corpus.txt')
```

## Procesamiento de corpus

Vamos a realizar algunas tareas que ya hemos resuelto, pero a través del uso de NLTK.

El fin de usar NLTK es comparar lo que hemos hecho con implementaciones dedicadas a la solución eficiente de problemas.

````python
with open ('file100.txt', 'r', encoding='utf8') as f:
    corpus = f.read()
print(corpus)
````

````python
#Vamos a usar la librería RE para eliminar los signos de puntuación y cualquier caracter no alfanumérico

corpus_clean1 = re.sub(r'[^\w\s]', ' ', corpus)    #regex para reemplazar caracteres no alfanuméricos por espacio
print(corpus_clean1)
````

````python
#Convertimos clean1 en LIST

corpus_clean1 = corpus_clean1.lower().split()
print(corpus_clean1)
````

````python
#Posteriormente, definimos una variable para que se guarden las palabras funcionales
#En esta variable activamos la stoplist de NLTK, en este caso, la de inglés

stopwords = set(stopwords.words('english'))

````

````python
print (stopwords)
````

````python
#Usamos los códigos que ya conocemos para eliminar las palabras funcionales

corpus_clean2 = (' '.join(w for w in corpus_clean1 if w not in stopwords)).split()
print (corpus_clean2)
````
