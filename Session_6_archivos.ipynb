Lo trabajado hasta ahora nos ha ejemplificado cómo trabajar con datos lingüísticos; sin embargo, no es viable trabajar con STR o LIST muy pequeños, sino con archivos en donde está guardado el corpus.

A contitnuación veremos cómo se trabajan los mismos procesos, pero a partir de archivos.

##Primero veremos una forma lógica, pero no muy funcional en algunos casos.

````python
##Primero, creamos una variable en donde guardaremos los datos del archivo.
##Ojo con el valor 'r': indica que lo abrimos con derechos de lectura, si quisiéramos guardar información (no recomendado en este momento),
##hay que cambiar la r por w

source = open('file1.txt', 'r', encoding = 'utf-8')
````

````python
##Luego creamos una segunda variable que lea los datos de source

#texto = source.read()          #carga el archivo como STR
#texto = source.readlines()     #carga el archivo como LIST
texto = source.readline(8)     #carga el archivo como STR y solo lee una línea

##Ya que hay una variable con esos datos, podemos ver el contenido de source

print(texto)
````

````python
##Es muy importante que después de abrir un archivo, se cierre, de lo contrario, Python sigue asignado memoria
source.close()
````

##Hay otras formas de trabajar con archivos que hacen que los procesos sean más eficientes y no sea necesario usar el método .close(). Además, con el método anterior se puede gastar mucha memoria si el archivo es muy grande.

A continuación, veremos la forma **with ... open**

````python
with open ('file1.txt', 'r', encoding = 'utf-8') as file:
    datos = file.read()
    print(datos)
````

````python
signos = '!"#$%&\'()*+,-.–/:;<=>?@[\]^_`{|}““~'
datos_clean1 = (''.join(s for s in datos if s not in signos)).lower()
print(datos_clean1)
````

````python
stoplist_EN = ['the', 'in', 'of', 'for', 'with', 'this', 'on', 'and', 'to', 'at', 'com', 'by', 'these', 'a']
datos_clean2 = datos_clean1.split()
#print(datos_clean2)
datos_clean3 = (' '.join(w for w in datos_clean2 if w not in stoplist_EN)).split()
print(datos_clean3)
````

````python
datos_fin = sorted(datos_clean3)
print(datos_fin)
````

````python
print(len(datos_fin), 'palabras en el corpus')
````

````python
tipos = {}
for i in datos_fin:
  tipos[i] = tipos.get(i, 0) + 1
print(tipos)
````

````python
with open ('salida.txt', 'w', encoding = 'utf-8') as salida:
  for i, toks in tipos.items():
    salida.write(i + ' ' + str(toks) + '\n')
    #print(i, toks)
````

Por ello, sólo veremos un ejemplo de cómo hacerlo y luego daremos paso a otra solución.


````python
import os
import glob
````

````python
files = glob.glob('*.txt')
for file in files:
  with open (file, 'r', encoding = 'utf-8') as file:
    allfiles = file.read()
  print(len(allfiles))

````
