# PANDAS: Introducción 🐼

PANDAS es una librería que permite la manipulación de diferentes tipos de datos de manera eficiente. La estructura de la librería hace que algunos de los procesos que hemos realizado hasta ahora sean más inmediatos en su codificación. Asimismo, hace que la estructuración y tratamiento de los datos sea muy flexible, lo cual facilita el análisis.

La documentación de la librería la encuentran en esta [liga](https://pandas.pydata.org).

Tal como lo hemos hecho con las librerías que hemos visto en el curso, lo primero que hay que hacer para usar la librería es instalarla:

## Estructura de documentos

PD trabaja con una estructura de datos muy semejante a Excel. Los datos se cargan en un *dataframe* (df), el cual tiene una organización bidimensional de los datos.

Esta forma de estructurar los datos es sumamente eficiente ya que permite trabajar con matrices de información muy amplias sin la complejidad de estar moviéndonos entre muchas filas y columnas. Pueden notar que, al procesar nuestros datos con PD, un dato textual se puede convertir en una representación numérica. Esto lo podíamos hacer mediante transformaciones con los procesos que hicimos en la primera parte del curso, pero ahora con PD el proceso es inmediato. Ello nos permite hacer procesos más descriptivos y visuales, es decir, podemos obtener información estadística, así como graficar lo que está sucediendo con la realidad que representan nuestros datos.

Para ejemplificar lo anterior, vamos a empezar a graficar algunos comportamientos de este conjunto de datos.

````python
corpus.NIVEL.value_counts().plot(kind='bar')
````

````python
corpus.ENTIDAD.value_counts().plot(kind='line',
                                   ls='dashed',
                                   title='SNIs por estado',
                                   color='green',
                                   figsize=(10, 6),
                                   grid=True,
                                   fontsize=6,
                                   xlabel=' Entidad',
                                   ylabel='Número de SNIs')
````
