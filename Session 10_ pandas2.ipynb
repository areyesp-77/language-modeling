##Buscando información de forma puntual

Métodos .iloc() y .loc()

Complementando los métodos .head() y .tail(), este par de métodos permiten hacer la búsqueda líneas específicas en el DF, por ejemplo, .iloc permite colocar la línea que necesito y recuperarla junto con su cabeceras asociadas.

El .loc opera de manera similar, solo que, en lugar de un número, buscas por etiquetas.

````python
df.iloc[3858]           #i = número, por ejemplo, si quiero un línea específica en lugar de .head o .tail
````

````python
df.iloc[3858:3862]
````

````python
df.loc[df.label == 1]           #El resultado complementa el método .value_counts
````

##Inserción y borrado

Podemos modificar el DF y añadir o borrar columnas.

A continuación veremos cómo añadimos una nueva columna llamada **POS** al DF. Para ello, vamos a duplicar los datos del DF original para evitar pérdida de datos.

````python
df1 = df
````

Posteriormente, añadimos una columna nueva en la última posición del DF.
Para el ejemplo, la columna nueva se llamará **agent** y el valor que le asignaremos será el de **human**.

````python
df1['agent'] = 'human'
df1
````

Ahora vamos a usar la información del DF para usar un método que, de forma automática, asigne dos valores diferentes a cada una de las celdas de la nueva columna.

La idea, recordando las secuencia de control, es la siguiente:



```
if df1.label == 1:
  POS.iloc[] = 'human'
else:
  POS.iloc[] = 'ai'
```



````python
df1['agent'] = np.where(df1.label == 1, 'human', 'ai')    #np.where() es un método de Numpy que funciona como if...else, solo que lo simplifica
df1
````

````python
df1['test'] = 'test'
df1
````

````python
df1.drop('test', axis=1, inplace=True)       #axis=1 indica que se busca en las columnas, inplace indica que el cambio es sobre el DF, de lo contrario, PD genera un DF nuevo
df1
````

## Otras formas de graficar

Con la librería **Seaborn (SNS)** se hacer graficación de forma similar a lo que hemos visto. El tipo de gráficos tiene una calidad semejante, pues SNS se apoya de Matplotlib.

En este [link](https://seaborn.pydata.org/tutorial/introduction.html#a-high-level-api-for-statistical-graphics) está la documentación inicial sobre la forma de graficar con SNS.

````python
# Gráfico tomado de la documentación de SNS.

fmri = sns.load_dataset("fmri")
sns.relplot(
    data=fmri, kind="line",
    x="timepoint", y="signal", col="region",
    hue="event", style="event",
)
````
