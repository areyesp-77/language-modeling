````python
#Preprocesamiento

import nltk
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')


from nltk import RegexpTokenizer
from nltk.corpus import stopwords
from nltk.tokenize import regexp_tokenize

import re
import string

with open ('toxic_spans.txt', 'r', encoding='utf8') as f:
    corpus = f.read()
#print(corpus)

corpus_clean1 = re.sub(r'[^\w\s]', ' ', corpus)
#print(corpus_clean1)

corpus_clean1 = corpus_clean1.lower().split()
#print(corpus_clean1)

freqs = {}
for word in corpus_clean1:
    freqs[word] = freqs.get(word, 0) + 1

with open ('results.txt', 'w', encoding = 'utf-8') as salida:
  for i, toks in freqs.items():
    salida.write(i + ' ' + str(toks) + '\n')

corpus_tagged = nltk.pos_tag(corpus_clean1)
#print(corpus_tagged)

with open ('pos_tags.txt', 'w', encoding='utf8') as f:
    for word, tag in corpus_tagged:
        f.write(f"{word}\t{tag} " + '\n')
````

````python
#Similitud contextual

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
corpus_lemmas = [lemmatizer.lemmatize(w) for w in corpus_clean1]

x = corpus_lemmas.count('muslim')
y = corpus_lemmas.count('people')
z = corpus_lemmas.count('woman')
print('Suma de KWs =', x+y+z)

freqs_lemmas = {}
for word in corpus_lemmas:
    freqs_lemmas[word] = freqs_lemmas.get(word, 0) + 1

with open ('results_lemmas.txt', 'w', encoding = 'utf-8') as salida:
  for i, toks in freqs_lemmas.items():
    salida.write(i + ' ' + str(toks) + '\n')

corpus_concord = nltk.Text(corpus_lemmas)
#corpus_concord.concordance('moron', 100, lines=30)

corpus_concord.similar('muslim')
````

````python
#Bonus

corpus_replace = []
for i in corpus_lemmas:
  if i == 'muslim':
    corpus_replace.append(1)
  elif i == 'people':
    corpus_replace.append(1)
  elif i == 'woman':
    corpus_replace.append(1)
  else:
    corpus_replace.append(0)
#print(corpus_replace)

print('Tamaño de N =', len(corpus_replace))

j = sum(corpus_replace)
print('Suma de Kws =', j)

k = j/len(corpus_replace)
print('Proporción de KWs/N =', k)
````
